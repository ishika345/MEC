```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score, mean_absolute_error
```

```python
df = pd.read_csv("synthetic_mec_data.csv")
```

```python
REQUIRED_COLUMNS = ["Temperature", "pH", "COD", "Voltage", "Current", "H2_Production"]
#missing_col = [c for c in req_columns if df[c].isnull().any()]
missing_col = [c for c in REQUIRED_COLUMNS if c not in df.columns ]
if missing_col:
  raise ValueError(" Uploaded file is missing coloumns: {missing_col}")

```

```python
interpolate_gap = 3
def interpolate_short_gaps(series, max_gap):
  is_na = series.isna()
  # label each consecutive run of NaNs with an id
  run_id = (is_na != is_na.shift()).cumsum()
  run_length = is_na.groupby(run_id).transform("sum")
  eligible = is_na &(run_id <= max_gap)
  interpolated = series. interpolate(method = 'linear', limit_dir = 'both')
  result = series.copy()
  result[eligible] = interpolated[eligible]
  return result
missing_counts = df[REQUIRED_COLUMNS].isna().sum()
missing_counts = missing_counts[missing_counts> 0]

```

```python
if len(missing_counts) > 0:
  print("Handling missing values (short gaps -> interpolate, long gaps -> drop):")
  for col in missing_counts.index:
    before_na = df[col].isna.sum()
    df[col] = interpolate_short_gaps(df[col], interpolate_gap)
    after_na = df[col].isna().sum()
    filled = before_na - after_na
    print(f"  {col}: {before_na} missing -> interpolated {filled}, {after_na} left in short gaps")
    print()

    n_before = len(df)
    df = df.dropna(subset=REQUIRED_COLUMNS)  # remaining NaNs = long gaps, drop those rows
    n_after = len(df)
    if n_after < n_before:
      print(f"Dropped {n_before - n_after} rows that fell inside gaps longer than {MAX_GAP_TO_INTERPOLATE} consecutive rows.\n")

    else:
      np.random.seed(42)
      n_samples = 500

      temperature = np.random.normal(loc=30, scale=1.5, size=n_samples)
    ph = np.random.normal(loc=7.0, scale=0.4, size=n_samples)
    cod = np.random.normal(loc=500, scale=30, size=n_samples)
    voltage = np.random.uniform(low=0.4, high=1.0, size=n_samples)
    current = 1.6 * voltage + np.random.normal(0, 0.05, n_samples)
    ph_penalty = -0.3 * (ph - 7.0) ** 2

    h2_production = (
        1.5 * current
        + 0.4 * voltage
        + 0.002 * cod
        + 0.05 * (temperature - 25)
        + ph_penalty
        + np.random.normal(0, 0.1, n_samples)
    )
    h2_production = np.clip(h2_production, a_min=0, a_max=None)

    df = pd.DataFrame({
        "Temperature": temperature,
        "pH": ph,
        "COD": cod,
        "Voltage": voltage,
        "Current": current,
        "H2_Production": h2_production,
    })
    print("Generated synthetic dataset (no real file loaded).")

print("First 5 rows:")
print(df.head())
print(f"\nTotal rows: {len(df)}\n")


```

```python
feature_col = ["Temperature", "pH", "COD", "Voltage", "Current"]
x= df[feature_col]
y = df["H2_Production"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(f" traing data: {len(x_train)} | testing data: {len(x_test)}\n")

```

```python
lr_model = LinearRegression()
lr_model.fit(x_train, y_train)
lr_y_pred = lr_model.predict(x_test)
lr_mse = mean_squared_error(y_test, lr_y_pred)
lr_rmse = np.sqrt(lr_mse)
lr_mae = mean_absolute_error(y_test, lr_y_pred)
lr_r2 = r2_score(y_test, lr_y_pred)
print(f"Mean Squared Error: {lr_mse:.2f}")
print(f"Root Mean Squared Error: {lr_rmse:.2f}")
print(f"Mean Absolute Error: {lr_mae:.2f}")
print(f"R-squared: {lr_r2:.2f}")
```

```python
for feature, coef in zip(feature_col, lr_model.coef_):
  print(f"{feature:<12}: {coef:.2f}")
print(f"{'Intercept':<12}: {lr_model.intercept_:.3f}")
```

```python
import matplotlib.pyplot as plt

labels = feature_col + ["Intercept"]
values = list(lr_model.coef_) + [lr_model.intercept_]

colors = ["skyblue" if v >= 0 else "orange" for v in values]

plt.bar(labels, values, color=colors)

plt.xlabel("Features")
plt.ylabel("Coefficients")
plt.title("Feature Coefficients + Intercept")

for i, v in enumerate(values):
    plt.text(
        i,
        v,
        f"{v:.3f}",
        ha="center",
        va="bottom" if v >= 0 else "top"
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
colors = [
    'skyblue' if abs(y_test.iloc[i] - lr_y_pred[i]) < 0.1 else 'orange'
    for i in range(len(y_test))
]
plt.scatter(y_test, lr_y_pred, color=colors)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], 'k--', lw=2, label = "perfect prediction")
plt.xlabel("Actual H2 Production")
plt.ylabel("Predicted H2 Production")
plt.title("Actual vs. Predicted H2 Production")
plt.legend()
plt.show()
```

```python
rf_model = RandomForestRegressor(n_estimators=200, max_depth = None,  random_state=42)
rf_model.fit(x_train, y_train)
rf_y_pred = rf_model.predict(x_test)
rf_mse = mean_squared_error(y_test, rf_y_pred)
rf_rmse = np.sqrt(rf_mse)
rf_mae = mean_absolute_error(y_test, rf_y_pred)
rf_r2 = r2_score(y_test, rf_y_pred)
print(f"Mean Squared Error: {rf_mse:.2f}")
print(f"Root Mean Squared Error: {rf_rmse:.2f}")
print(f"Mean Absolute Error: {rf_mae:.2f}")
print(f"R-squared: {rf_r2:.2f}")
```

```python
xgbmodel = XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
xgbmodel.fit(x_train, y_train)
xgb_y_pred = xgbmodel.predict(x_test)
xgb_mse = mean_squared_error(y_test, xgb_y_pred)
xgb_rmse = np.sqrt(xgb_mse)
xgb_mae = mean_absolute_error(y_test, xgb_y_pred)
xgb_r2 = r2_score(y_test, xgb_y_pred)
print(f"Mean Squared Error: {xgb_mse:.2f}")
print(f"Root Mean Squared Error: {xgb_rmse:.2f}")
print(f"Mean Absolute Error: {xgb_mae:.2f}")
print(f"R-squared: {xgb_r2:.2f}")

```

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest", "XGBoost"],
    "MAE": [lr_mae, rf_mae, xgb_mae],
    "RMSE": [lr_rmse, rf_rmse, xgb_rmse],
    "R2": [lr_r2, rf_r2, xgb_r2],
}).round(3)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))
colors = ["skyblue", "lightgreen", "orange"]  # one consistent color per model, across all 3 panels

metrics = ["MAE", "RMSE", "R2"]
ylabels = ["MAE (L/day)", "RMSE (L/day)", "R²"]

# zip() pairs each subplot axis with its metric column and y-axis label,
# so this one loop draws all three panels instead of writing 3 near-identical blocks
for ax, metric, ylabel in zip(axes, metrics, ylabels):
    bars = ax.bar(comparison["Model"], comparison[metric], color=colors)
    ax.set_title(metric)
    ax.set_ylabel(ylabel)
    ax.tick_params(axis="x", rotation=20)
    for bar, val in zip(bars, comparison[metric]):
        ax.text(bar.get_x() + bar.get_width() / 2, val, f"{val:.3f}", ha="center", va="bottom")

fig.suptitle("Model Comparison: Linear Regression vs Random Forest vs XGBoost")
plt.tight_layout()
plt.show()
```
